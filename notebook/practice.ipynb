```python
def contains_duplicates(lst): 
    hash_set = set()
    for num in lst:
        if num in hash_set:
            return True
        else:
            hash_set.add(num)
    return False

print(contains_duplicates(lst))
```

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def display_info():
        print(f"Car Information: {self.year} {self.make} {self.model}")

my_car = Car("Toyota", "Camry", 2020)
my_car.display_info()
```
